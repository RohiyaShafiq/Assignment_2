```python
from common.data import read_jsonl, prompt_messages_from_preference
from common.models import load_tokenizer

rows = read_jsonl("data/dpo_standard_train.jsonl")

tokenizer = load_tokenizer("Qwen/Qwen2.5-1.5B-Instruct")

bad = []

for i, row in enumerate(rows):
    messages = prompt_messages_from_preference(row)

    prompt_ids = tokenizer.apply_chat_template(
        messages,
        tokenize=True,
        add_generation_prompt=True,
    )

    if len(prompt_ids) >= 768:
        bad.append((i, len(prompt_ids)))

print("Total examples:", len(rows))
print("Examples with prompt >= 768:", len(bad))
print("First 20:", bad[:20])
print("Maximum prompt length:", max(x[1] for x in bad) if bad else 0)
```
